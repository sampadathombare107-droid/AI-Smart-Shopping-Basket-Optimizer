```python
rules = association_rules(
    frequent_items,
    metric="confidence",
    min_threshold=0.1
)

rules = rules.sort_values(by="confidence", ascending=False)

rules[
    [
        "antecedents",
        "consequents",
        "support",
        "confidence",
        "lift"
    ]
].head(10)
```

```python
def recommend(product_name):
    result = rules[
        rules["antecedents"].apply(
            lambda x: product_name in list(x)
        )
    ]

    return result[
        ["consequents", "confidence", "lift"]
    ].sort_values(
        by="confidence",
        ascending=False
    )
```

```python
recommend("yogurt")
```
